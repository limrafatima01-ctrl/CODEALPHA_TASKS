# Task 3: Data Visualization

## Objective

The objective of this task is to transform e-commerce data into meaningful visualizations using Python libraries such as Matplotlib and Seaborn. The visualizations will help identify sales trends, category performance, city-wise orders, and other useful patterns in the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../web_scraping/scraped_books_dataset.csv")

df.head()
```

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Number"] = df["Rating"].map(rating_map)

df["Rating_Number"].value_counts().sort_index()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Rating_Number", color="steelblue")

plt.title("Distribution of Book Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Price", bins=10, kde=True, color="orange")

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.tight_layout()
plt.show()
```

```python
top_books = df.nlargest(10, "Price")[["Title", "Price"]].sort_values("Price")

plt.figure(figsize=(10, 6))

sns.barplot(data=top_books, x="Price", y="Title", color="teal")

plt.title("Top 10 Most Expensive Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Rating_Number", y="Price", color="lightblue")

plt.title("Book Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.tight_layout()
plt.show()
```

## Key Insights

- The dataset contains 20 books collected from a public webpage.
- Book ratings range from 1 to 5.
- The price histogram shows the distribution of book prices.
- The top 10 chart makes it easy to compare the most expensive books.
- The box plot helps compare book prices across different rating levels.
- Data visualization makes the scraped dataset easier to understand and interpret.

## Conclusion

The scraped book dataset was transformed into meaningful visualizations using Matplotlib and Seaborn. The charts helped visualize book ratings, price distribution, expensive books, and the relationship between ratings and prices.

This task demonstrated how data visualization can make structured data easier to understand and communicate.
